# **"Phase 4 — Enrichment"**

```python
def experience_level(years_code):
    if pd.isna(years_code):
        return "Unknown"
    if years_code <= 2:
        return "Beginner"
    if years_code <= 5:
        return "Early Career"
    if years_code <= 10:
        return "Experienced"
    return "Highly Experienced"
```

```python
df["ExperienceLevel"] = df["YearsCode"].apply(experience_level)
```

```python
df["ExperienceLevel"].value_counts()
```

```python
df["LearnCode"].explode().unique()
```

```python
def uses_documentation(learn_code_list):
    if type(learn_code_list) is not list:
        return False
    return "Technical documentation (is generated for/by the tool or system)" in learn_code_list

df["usesDocumentation"] = df["learnCode"].apply(uses_documentation)
```

```python
def uses_ai_for_learning(learn_code_list):
    if type(learn_code_list) is not list:
        return False
    return "AI CodeGen tools or AI-enabled apps" in learn_code_list

df["usesAIForLearning"] = df["LearnCode"].apply(uses_ai_for_learning)
```

```python
def uses_stack_overflow(learn_code_list):
    if type(learn_code_list) is not list:
        return False
    return "Stack Overflow or Stack Exchange" in learn_code_list

df["usesStackOverflow"] = df["learningMethods"].apply(uses_stack_overflow)
```

```python
df["usesStackOverflow"].value_counts()
```

```python
df = df.rename(columns={
    'ResponseId': 'responseId',
    'Age': 'age',
    'YearsCode': 'yearsCode',
    'DevType': 'devType',
    'LearnCodeChoose': 'learnCodeChoose',
    'LearnCode': 'learningMethods',
    'LearnCodeAI': 'learnCodeAI',
    'AILearnHow': 'aiLearningMethods',
    'AISelect': 'aiUsage',
    'AIAcc': 'aiTrust',
    'AISent': 'aiSentiment',
    "ExperienceLevel" : "experienceLevel"
})
```
